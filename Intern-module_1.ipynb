# 3. Python Operators and Input/Output

Operators are symbols used to perform operations on variables and values.
Python also allows users to enter data using input() and display results using print().

```python
# Arithmetic Operators

a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

```python
# Comparison Operators

a = 10
b = 5

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
#user input and output

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Hello", name)
print("Your age is", age)
```

## 4. Conditional Statements

Conditional statements are used to make decisions based on conditions.

Python supports if, elif and else statements.

```python

marks = 75

if marks >= 40:
    print("You are Pass")
```

```python

marks = 30

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

```python

marks = 85

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

```python
sales = 15000

if sales >= 20000:
    print("High Sales")
elif sales >= 10000:
    print("Medium Sales")
else:
    print("Low Sales")
```

## 5. Loops in Python

Loops are used to execute a block of code repeatedly.

Python provides two main types of loops:
1. for loop
2. while loop

```python

for i in range(1, 6):
    print(i)
```

```python
#Print numbers from 1 to 10
for i in range(1, 11):
    print(i)
```

```python
#Calculate the sum of numbers

total = 0

for i in range(1, 6):
    total = total + i

print("Total:", total)
```

```python
#Loop Through a List

sales = [1200, 2500, 1800, 3200, 1500]

for amount in sales:
    print(amount)
```

```python
#The while Loop

i = 1

while i <= 5:
    print(i)
    i = i + 1
```

```python
#Real-World Example: Sales Analysis

sales = [1200, 2500, 1800, 3200, 1500]

total_sales = 0

for amount in sales:
    total_sales = total_sales + amount

average_sales = total_sales / len(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
```

## 6. Lists and Dictionaries

Lists are used to store multiple items in a single variable.
Dictionaries store data in key-value pairs.

```python
# Creating a list
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits)
print(type(fruits))
```

```python

fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[-1])
```

```python

fruits = ["Apple", "Banana", "Mango"]

# Add an item
fruits.append("Orange")
print("After adding:", fruits)

# Update an item
fruits[1] = "Grapes"
print("After updating:", fruits)

# Remove an item
fruits.remove("Apple")
print("After removing:", fruits)
```

```python

sales = [1200, 2500, 1800, 3200, 1500]

print("Total:", sum(sales))
print("Maximum:", max(sales))
print("Minimum:", min(sales))
print("Count:", len(sales))
print("Average:", sum(sales) / len(sales))
```

```python

student = {
    "name": "Shubham",
    "age": 21,
    "marks": 85
}

print(student)
print(student["name"])
print(student["age"])
print(student["marks"])
```

```python

student = {
    "name": "Shubham",
    "age": 21,
    "marks": 85
}

# Add a new key-value pair
student["course"] = "Data Science"

# Update a value
student["marks"] = 90

# Remove a key-value pair
student.pop("age")

print(student)
```

```python
# Real-World Example: Product Sales

products = [
    {"name": "Laptop", "sales": 50000},
    {"name": "Mobile", "sales": 30000},
    {"name": "Tablet", "sales": 20000}
]

total_sales = 0

for product in products:
    print(product["name"], "-", product["sales"])
    total_sales += product["sales"]

print("Total Sales:", total_sales)
```

## 7. Functions in Python

A function is a reusable block of code that performs a specific task.

Functions help make programs organized, reusable and easier to maintain.

```python
#simple function
def greet():
    print("Welcome to Data Science!")

greet()
```

```python
#Function with Parameters
def greet(name):
    print("Hello", name)

greet("Shubham")
greet("Rahul")
```

```python
#Function with Return Value
def add(a, b):
    return a + b

result = add(10, 20)

print("Sum:", result)
```

```python
#Real-World Example: Sales Analysis

def calculate_total(sales):
    return sum(sales)

def calculate_average(sales):
    return sum(sales) / len(sales)

sales = [1200, 2500, 1800, 3200, 1500]

print("Total Sales:", calculate_total(sales))
print("Average Sales:", calculate_average(sales))
```

## 8. File Handling in Python

File handling allows Python to create, read, write and manage files.

Common file modes:
- r: Read
- w: Write
- a: Append

```python
with open("sales.txt", "w") as file:
    file.write("Laptop Sales: 50000\n")
    file.write("Mobile Sales: 30000\n")

print("File created successfully!")
```

```python
# Reading a File
with open("sales.txt", "r") as file:
    content = file.read()

print(content)
```

```python
#Appending Data

with open("sales.txt", "a") as file:
    file.write("Tablet Sales: 20000\n")

print("New data added!")
```

# 9. Data Science Concepts and Real-World Applications

## What is Data Science?

Data Science is a field that uses data, statistics, programming and analytical techniques to extract useful information and insights from data.

Python is widely used in Data Science for data cleaning, data analysis, visualization and machine learning.

## Real-World Applications of Data Science

1. Healthcare: Predicting diseases and analyzing medical data.
2. Banking: Detecting fraudulent transactions.
3. E-commerce: Recommending products to customers.
4. Retail: Analyzing sales and customer behavior.
5. Transportation: Predicting traffic and optimizing routes.
6. Education: Analyzing student performance.

## Data Science Process

1. Data Collection
2. Data Cleaning
3. Data Analysis
4. Data Visualization
5. Model Building
6. Decision Making

```python
#Simple Data Science Example

sales = [1200, 2500, 1800, 3200, 1500]

total_sales = sum(sales)
avg_sales = total_sales/len(sales)
highest_sales = max(sales)

print("Sales Data:", sales)
print("Total Sales:",total_sales)
print("Average Sales:", avg_sales)
print("Highest Sales:", highest_sales)
```

### Practical Learning Outcome

In this example, I used Python to analyze simple sales data.

- Data Collection: Collected sales values.
- Data Analysis: Calculated total, average and highest sales.
- Decision Making: Identified the highest sales value.

This example demonstrates how Python can be used for basic data analysis and business decision-making.
